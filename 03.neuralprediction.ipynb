# Chapter 3 - Introduction to Neural Prediction: Forward Propagation

## Summary
Builds the first neural networks and performs **forward propagation** (making predictions).
- A **neuron** takes inputs, multiplies by **weights**, sums (weighted sum / dot product).
- Cases: single input -> single output, **multiple inputs**, **multiple outputs**, **multiple inputs & outputs**, and **hidden layers** (stacked networks).
- NumPy gives vectorised `dot`, replacing loops.

## Theory
- **Weight:** a "volume knob" on how much an input influences a prediction.
- **Weighted sum:** `pred = sum(input_i * weight_i)` = dot product. Sensitivity depends on weight sign/size.
- **Multiple outputs:** each output has its own weight vector; the weights form a **matrix**; computing is `vector @ matrix`.
- **Hidden layers:** `layer_1 = input @ W_0`, `pred = layer_1 @ W_1`. Stacked (without non-linearity) layers are just one linear map; non-linearity comes later (Ch6).

```python
import numpy as np

# 1) Single input, single output
def neural_network(inp, weight): return inp * weight
print("single:", neural_network(8.5, 0.1))
```

```python
# 2) Multiple inputs, single output (weighted sum)
def w_sum(a, b): return sum(x * y for x, y in zip(a, b))
weights = [0.1, 0.2, 0]
toes, wlrec, nfans = [8.5, 9.5, 9.9, 9.0], [0.65, 0.8, 0.8, 0.9], [1.2, 1.3, 0.5, 1.0]
inp = [toes[0], wlrec[0], nfans[0]]
print("multi-in:", w_sum(inp, weights))

# NumPy version
print("numpy dot:", np.array(inp).dot(np.array(weights)))
```

```python
# 3) Single input, multiple outputs
def ele_mul(n, v): return [n * x for x in v]
w = [0.3, 0.2, 0.9]
print("multi-out:", ele_mul(wlrec[0], w))
```

```python
# 4) Multiple inputs and outputs: vector @ matrix
weights = np.array([[0.1, 0.1, -0.3],   # hurt?
                    [0.1, 0.2,  0.0],   # win?
                    [0.0, 1.3,  0.1]])  # sad?
inp = np.array([toes[0], wlrec[0], nfans[0]])
print("multi-in/out:", inp.dot(weights))
```

```python
# 5) Hidden layer: stacked networks
ih_wgt = np.array([[0.1, 0.2, -0.1],
                   [-0.1, 0.1, 0.9],
                   [0.1, 0.4, 0.1]]).T
hp_wgt = np.array([[0.3, 1.1, -0.3],
                   [0.1, 0.2, 0.0],
                   [0.0, 1.3, 0.1]]).T
def stacked(inp, ih, hp):
    hid = ih.dot(inp)
    return hp.dot(hid)
print("stacked:", stacked(inp, ih_wgt, hp_wgt))
```

## Takeaways
Forward propagation = repeated dot products. Weights are the learnable parameters; the next chapter shows how to learn them.
