# Chapter 1 - Introducing Deep Learning: Why You Should Learn It

## Summary
Chapter 1 is conceptual. It motivates deep learning, explains why *understanding the internals* (rather than only calling a framework) matters, and sets expectations: you only need basic Python and high-school math.

Key ideas:
- **Deep learning** is a subset of **machine learning**, itself a subset of **AI**. It uses stacks of simple, differentiable functions (layers of artificial neurons) that learn from data.
- Deep learning is **"learning by example"**: instead of hand-writing rules, we show the machine inputs and desired outputs.
- It is a **toolbox for mimicking human-like perception/prediction** (vision, language, audio).
- The book teaches by building networks **from scratch with NumPy**, so nothing is a black box.
- Prerequisites: Python, a bit of linear algebra (vectors, dot products), and a Jupyter environment.

## Theory: Why learn it from scratch?
| Concept | Explanation |
|---|---|
| Machine learning | Algorithms that improve at a task via data rather than explicit programming. |
| Deep learning | ML using many-layered neural networks; each layer transforms its input into a more abstract representation. |
| Parametric model | A model with adjustable numbers (weights). Learning = finding good weights. |
| Supervised learning | Learn mapping input -> output from labelled examples. |
| Unsupervised learning | Find structure in unlabeled data (clustering, etc.). |

Understanding the internals lets you debug, tune and invent new architectures, rather than treat frameworks as magic.

## Environment check
Chapter 1 has no book code; here is a setup check and a tiny "learning by example" demo (fit y = 2x from examples).

```python
import sys, numpy as np
print(sys.version)
print("NumPy", np.__version__)
```

```python
# Tiny demo of learning from examples: discover the rule y = 2x
xs = np.array([1., 2., 3., 4.])
ys = np.array([2., 4., 6., 8.])
w = 0.0
for _ in range(50):
    for x, y in zip(xs, ys):
        w -= 0.01 * (x * w - y) * x   # nudge weight to reduce error
print("learned weight:", round(w, 4), "(true value: 2)")
```

## Takeaways
- Deep learning = learning weights from examples.
- The rest of the book builds the toolkit step by step: prediction (Ch3), error + gradient descent (Ch4-5), backpropagation (Ch6).
