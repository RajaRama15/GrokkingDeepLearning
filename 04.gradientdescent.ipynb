# Chapter 4 - Introduction to Neural Learning: Gradient Descent

## Summary
How a network *learns*: compare prediction to a goal, then adjust weights.
- **Error measures:** mean squared error `(pred - goal)^2` (always positive, penalises big misses).
- **Hot and cold learning:** try weight +/- a small amount, keep the direction that lowers error (simple but slow/imprecise).
- **Gradient descent:** use the **derivative** of error w.r.t. weight to move directly toward lower error.
- Core formula: `delta = pred - goal`, `weight_delta = delta * input`, `weight -= alpha * weight_delta`.
- **alpha** (learning rate) controls step size; too large causes divergence.

## Theory
Error `E = (w*x - y)^2`. Derivative: `dE/dw = 2 (w x - y) x`. The book drops the constant 2 (absorbed into alpha).
- **Direction:** sign of derivative says which way to move.
- **Amount:** magnitude scales with how wrong we are and with the input size.
- **Divergence:** if `alpha * x^2 > 1`, updates overshoot and error grows.

```python
# Hot and cold learning
weight, goal, inp, step = 0.5, 0.8, 0.5, 0.001
for it in range(1101):
    pred = inp * weight
    err = (pred - goal) ** 2
    up = (inp * (weight + step) - goal) ** 2
    down = (inp * (weight - step) - goal) ** 2
    if down < up: weight -= step
    else: weight += step
    if it % 275 == 0: print(f"iter {it}: error={err:.6f} pred={pred:.4f}")
```

```python
# Gradient descent
weight, goal, inp, alpha = 0.5, 0.8, 2.0, 0.1
for it in range(20):
    pred = inp * weight
    error = (pred - goal) ** 2
    delta = pred - goal
    weight_delta = delta * inp
    weight -= alpha * weight_delta
    print(f"Error: {error:.6f} Prediction: {pred:.6f}")
```

```python
# Divergence with a too-large alpha
weight, goal, inp, alpha = 0.5, 0.8, 2.0, 0.5
for it in range(6):
    pred = inp * weight
    print(f"Error: {(pred-goal)**2:.4f} Prediction: {pred:.4f}")
    weight -= alpha * (pred - goal) * inp
```

## Takeaways
Gradient descent = compute error slope and step against it. Learning rate must be tuned.
