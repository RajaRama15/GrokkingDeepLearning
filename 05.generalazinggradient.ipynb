# Chapter 5 - Learning Multiple Weights at a Time: Generalizing Gradient Descent

## Summary
Extends gradient descent to networks with **multiple inputs**, **multiple outputs**, and both.
- Multiple inputs: one shared `delta`; each weight gets `weight_delta[i] = input[i] * delta`.
- Multiple outputs: independent error/delta per output.
- Inputs -> outputs: weight matrix, `weight_delta = outer(input, delta)`.
- **Freezing a weight** shows weights learn *together* but with different speeds; error can still reach zero with a frozen weight.
- Weights visualisation reveals what the network focuses on (preview of interpretation).

## Theory
For a weighted sum `pred = sum(x_i w_i)`, the partial derivative of the error wrt `w_i` is `delta * x_i`, so inputs with larger values get larger updates. Because each weight is updated by its own partial derivative, all weights descend the same error surface at once. In the matrix case, `dE/dW = x^T (outer) delta`.

```python
import numpy as np
def neural_network(inp, weights): return np.array(inp).dot(weights)

toes, wlrec, nfans = [8.5, 9.5, 9.9, 9.0], [0.65, 0.8, 0.8, 0.9], [1.2, 1.3, 0.5, 1.0]
win_or_lose = [1, 1, 0, 1]

# Multiple inputs
weights = np.array([0.1, 0.2, -0.1])
inp = np.array([toes[0], wlrec[0], nfans[0]])
goal, alpha = win_or_lose[0], 0.01
for it in range(3):
    pred = neural_network(inp, weights)
    error = (pred - goal) ** 2
    delta = pred - goal
    weight_deltas = inp * delta
    weights -= alpha * weight_deltas
    print(f"iter {it}: error={error:.4f} pred={pred:.4f}")
```

```python
# Freezing one weight: the others compensate
weights = np.array([0.1, 0.2, -0.1])
for it in range(3):
    pred = neural_network(inp, weights)
    delta = pred - goal
    wd = inp * delta
    wd[0] = 0                      # freeze weight 0
    weights -= alpha * wd
    print(f"iter {it}: error={(pred-goal)**2:.4f}")
print(weights)
```

```python
# Multiple outputs
weights = np.array([0.3, 0.2, 0.9])
hurt, win, sad = [0.1, 0.0, 0.0, 0.1], [1, 1, 0, 1], [0.1, 0.0, 0.1, 0.2]
inp, goals, alpha = wlrec[0], np.array([hurt[0], win[0], sad[0]]), 0.1
for it in range(3):
    pred = inp * weights
    delta = pred - goals
    weights -= alpha * delta * inp
    print(f"iter {it}: error={(delta**2).sum():.4f}")
```

```python
# Multiple inputs AND outputs
weights = np.array([[0.1, 0.1, -0.3],
                    [0.1, 0.2,  0.0],
                    [0.0, 1.3,  0.1]])
inp = np.array([toes[0], wlrec[0], nfans[0]])
goals = np.array([hurt[0], win[0], sad[0]])
alpha = 0.01
for it in range(5):
    pred = inp.dot(weights)
    delta = pred - goals
    weights -= alpha * np.outer(inp, delta)
    print(f"iter {it}: error={(delta**2).sum():.4f}")
```

## Takeaways
`weight_delta = input x delta` generalises to any number of inputs/outputs; this is the same outer-product update used in all later chapters.
