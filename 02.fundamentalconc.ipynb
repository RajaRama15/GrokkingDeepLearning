# Chapter 2 - Fundamental Concepts: How Do Computers Learn?

## Summary
Introduces the learning paradigms and the intuition behind *predict -> compare -> learn*.

- **Paradigms:** supervised (labelled data), unsupervised (cluster/structure), reinforcement (reward signals).
- **Parametric vs. non-parametric:** parametric models have a fixed number of adjustable parameters (neural nets); non-parametric models (e.g. kNN) grow with data.
- **Supervised parametric learning** in 3 steps: **Predict**, **Compare** (measure error), **Learn** (adjust parameters).
- The chapter uses a *"dial/knob"* analogy: a machine takes input data, and knobs (weights) are turned to make predictions better.

## Theory
- **Predict:** `pred = f(input, weights)`.
- **Compare:** error measure, e.g. `(pred - goal)^2`. Tells us *how much* we missed.
- **Learn:** change weights to shrink error.
- **Unsupervised, e.g. clustering:** group points without labels.
- **Non-parametric (kNN):** prediction uses stored training examples directly - no training of weights.

```python
import numpy as np

# Supervised parametric learning: predict -> compare -> learn
x, goal, w, lr = 2.0, 10.0, 1.0, 0.05
for i in range(8):
    pred = x * w                  # PREDICT
    err = (pred - goal) ** 2      # COMPARE
    w -= lr * 2 * (pred - goal) * x  # LEARN
    print(f"step {i}: pred={pred:.3f} error={err:.3f} w={w:.3f}")
```

```python
# Non-parametric example: k-nearest-neighbours (k=3) on toy data
train_x = np.array([1, 2, 3, 10, 11, 12], float)
train_y = np.array([0, 0, 0, 1, 1, 1])
def knn(q, k=3):
    idx = np.argsort(abs(train_x - q))[:k]
    return np.bincount(train_y[idx]).argmax()
print(knn(2.5), knn(9.0))
```

```python
# Unsupervised: tiny 1-D k-means (k=2) discovers two groups without labels
pts = np.array([1, 2, 3, 10, 11, 12], float)
c = np.array([0., 5.])
for _ in range(10):
    lab = np.argmin(abs(pts[:, None] - c[None]), axis=1)
    c = np.array([pts[lab == j].mean() for j in range(2)])
print("centres:", c, "labels:", lab)
```

## Takeaways
Neural nets are parametric, supervised learners; the loop **predict -> compare -> learn** is the engine of the entire book.
